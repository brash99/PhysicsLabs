# Physical model

Let $m_c$ be the total moving cart mass (cart, flag, and added cart masses) and $m_h$ the total hanging mass (hanger plus added masses). Neglecting friction, pulley inertia, string mass, and air drag, Newton's second law gives

$$
m_hg-T=m_ha,\qquad T=m_ca.
$$

Therefore,

$$
a_{\rm pred}=\frac{m_hg}{m_c+m_h},
\qquad
T_{\rm pred}=m_ca_{\rm pred}=\frac{m_cm_hg}{m_c+m_h}.
$$

For the cart alone, $T=m_ca$. A graph of predicted tension $T_{\rm pred}$ versus measured acceleration $a_{\rm meas}$ should therefore be linear. Its slope should agree with $m_c$, and its intercept should be consistent with zero for an ideal track. A nonzero intercept can reveal a nearly constant resistive force or a small track tilt.

The acceleration measured with two photogates is

$$
a_{\rm meas}=\frac{v_2-v_1}{t_2-t_1}.
$$

This model assumes constant acceleration between the gates. Write down which assumptions you can test directly and which remain idealizations.


# Uncertainty plan

Use standard uncertainties and combine independent contributions in quadrature.

For one trial, with $\Delta v=v_2-v_1$ and $\Delta t=t_2-t_1$,

$$
u_a^2=\left(\frac{u_{v1}}{\Delta t}\right)^2+
\left(\frac{u_{v2}}{\Delta t}\right)^2+
\left(\frac{\Delta v\,u_{t1}}{\Delta t^2}\right)^2+
\left(\frac{\Delta v\,u_{t2}}{\Delta t^2}\right)^2.
$$

For repeated trials, the notebook combines the standard error of the trial accelerations with the propagated measurement floor. For each mass configuration it also propagates the mass and $g$ uncertainties into $a_{\rm pred}$ and $T_{\rm pred}$.

Compare measured and predicted acceleration with

$$
z=\frac{|a_{\rm meas}-a_{\rm pred}|}{\sqrt{u_{a,\rm meas}^2+u_{a,\rm pred}^2}}.
$$

As a practical guide, $z\lesssim2$ is consistent within combined uncertainty. A larger value deserves a physical explanation; it is not a reason to delete trials.

## Validate and calculate every trial

The next cell checks common entry problems before calculating acceleration. If a trial is objectively rejected, preserve it in `raw`, explain the reason in `note`, and add its row number to `rejected_rows`. The default is to retain every row.


```python
required = {
    "cart_set", "trial", "m_cart_kg", "u_m_cart_kg", "m_hang_kg",
    "u_m_hang_kg", "t1_s", "v1_m_s", "t2_s", "v2_m_s", "note"
}
missing = required - set(raw.columns)
assert not missing, f"Missing columns: {sorted(missing)}"
assert raw[list(required - {"cart_set", "note"})].notna().all().all(), "Numeric data contain blanks."
assert (raw["t2_s"] > raw["t1_s"]).all(), "Every t2 must be greater than t1."
assert (raw["v2_m_s"] > raw["v1_m_s"]).all(), "Every v2 should exceed v1 for this setup."
assert (raw[["m_cart_kg", "m_hang_kg", "u_m_cart_kg", "u_m_hang_kg"]] > 0).all().all()

# Example: rejected_rows = [12] after documenting the objective reason in raw.loc[12, "note"]
rejected_rows = []
analysis = raw.drop(index=rejected_rows).copy()

analysis["delta_t_s"] = analysis["t2_s"] - analysis["t1_s"]
analysis["delta_v_m_s"] = analysis["v2_m_s"] - analysis["v1_m_s"]
analysis["a_meas_m_s2"] = analysis["delta_v_m_s"] / analysis["delta_t_s"]
analysis["u_a_trial_m_s2"] = np.sqrt(
    (u_v / analysis["delta_t_s"])**2
    + (u_v / analysis["delta_t_s"])**2
    + (analysis["delta_v_m_s"] * u_t / analysis["delta_t_s"]**2)**2
    + (analysis["delta_v_m_s"] * u_t / analysis["delta_t_s"]**2)**2
)

analysis[[
    "cart_set", "trial", "m_hang_kg", "delta_t_s", "delta_v_m_s",
    "a_meas_m_s2", "u_a_trial_m_s2", "note"
]]
```

## Summarize each mass configuration

For repeated accelerations, the code calculates the sample standard deviation and standard error. It combines the standard error with the propagated instrument floor for the mean. It then calculates the ideal prediction and its uncertainty using partial derivatives and quadrature.


```python
keys = ["cart_set", "m_cart_kg", "u_m_cart_kg", "m_hang_kg", "u_m_hang_kg"]
summary = (
    analysis.groupby(keys, as_index=False)
    .agg(
        n=("a_meas_m_s2", "size"),
        a_meas_m_s2=("a_meas_m_s2", "mean"),
        s_a_m_s2=("a_meas_m_s2", "std"),
        u_a_floor_m_s2=("u_a_trial_m_s2", lambda x: np.sqrt(np.sum(x**2)) / len(x)),
    )
)
summary["sem_a_m_s2"] = summary["s_a_m_s2"] / np.sqrt(summary["n"])
summary["u_a_meas_m_s2"] = np.sqrt(summary["sem_a_m_s2"]**2 + summary["u_a_floor_m_s2"]**2)

mc = summary["m_cart_kg"]
mh = summary["m_hang_kg"]
umc = summary["u_m_cart_kg"]
umh = summary["u_m_hang_kg"]
total = mc + mh

summary["a_pred_m_s2"] = mh * g / total
summary["u_a_pred_m_s2"] = np.sqrt(
    (-mh * g / total**2 * umc)**2
    + (mc * g / total**2 * umh)**2
    + (mh / total * u_g)**2
)

summary["T_pred_N"] = mc * mh * g / total
summary["u_T_pred_N"] = np.sqrt(
    (mh**2 * g / total**2 * umc)**2
    + (mc**2 * g / total**2 * umh)**2
    + (mc * mh / total * u_g)**2
)

summary["z_acceleration"] = (
    (summary["a_meas_m_s2"] - summary["a_pred_m_s2"]).abs()
    / np.sqrt(summary["u_a_meas_m_s2"]**2 + summary["u_a_pred_m_s2"]**2)
)

summary[[
    "cart_set", "m_cart_kg", "m_hang_kg", "n", "a_meas_m_s2",
    "u_a_meas_m_s2", "a_pred_m_s2", "u_a_pred_m_s2",
    "T_pred_N", "u_T_pred_N", "z_acceleration"
]]
```

# Fit the force–acceleration relationship

The orthogonal-distance regression below accounts for uncertainty in both measured acceleration (horizontal axis) and predicted tension (vertical axis). For each cart setting it fits

$$
T=b+Ma.
$$

Compare the fitted slope $M$ with the independently measured cart mass $m_c$. Also interpret the intercept $b$ physically.


```python
def line(beta, x):
    return beta[0] + beta[1] * x


fit_rows = []
fig, ax = plt.subplots(figsize=(8, 6))
colors = plt.cm.tab10.colors

for color, (cart_set, group) in zip(colors, summary.groupby("cart_set", sort=False)):
    x = group["a_meas_m_s2"].to_numpy()
    sx = group["u_a_meas_m_s2"].to_numpy()
    y = group["T_pred_N"].to_numpy()
    sy = group["u_T_pred_N"].to_numpy()
    measured_mass = group["m_cart_kg"].iloc[0]
    u_measured_mass = group["u_m_cart_kg"].iloc[0]

    data = odr.RealData(x, y, sx=sx, sy=sy)
    result = odr.ODR(data, odr.Model(line), beta0=[0.0, measured_mass]).run()
    intercept, slope = result.beta
    u_intercept, u_slope = result.sd_beta
    z_slope = abs(slope - measured_mass) / np.sqrt(u_slope**2 + u_measured_mass**2)

    fit_rows.append({
        "cart_set": cart_set,
        "measured_cart_mass_kg": measured_mass,
        "u_measured_cart_mass_kg": u_measured_mass,
        "fit_slope_kg": slope,
        "u_fit_slope_kg": u_slope,
        "fit_intercept_N": intercept,
        "u_fit_intercept_N": u_intercept,
        "z_slope_vs_mass": z_slope,
        "reduced_chi_square": result.res_var,
    })

    ax.errorbar(x, y, xerr=sx, yerr=sy, fmt="o", capsize=3, color=color, label=f"{cart_set} data")
    xfit = np.linspace(max(0, 0.9*x.min()), 1.08*x.max(), 150)
    ax.plot(xfit, line(result.beta, xfit), color=color, label=f"{cart_set} ODR fit")

ax.set_xlabel(r"Measured acceleration $a$ (m/s$^2$)")
ax.set_ylabel(r"Predicted tension $T$ (N)")
ax.set_title("Figure 1. Tension versus measured acceleration")
ax.grid(alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()

fit_results = pd.DataFrame(fit_rows)
fit_results
```
